## 2. Data Loading: Web Scraping Pipeline

VGChartz paginates its Game Database using the `page` query parameter (`page=1` - `page=5`). Each page has a main HTML table inside `<div id="generalBody">`.

Helper functions below:
- `parse_table_headers`: Maps each column header name to its cell index, accounting for `colspan` attributes.
- `scrape_vgchartz_page`: Downloads one database page with a `User-Agent` header and parses every game row, extracting console abbreviations from `<img alt="...">` tags.

```python
def parse_table_headers(table):
    """Extract column header names from a VGChartz table, expanding colspan attributes."""
    header_row = None
    for row in table.find_all("tr"):
        th_cells = row.find_all("th")
        # Identify the main header row containing 'Game' and 'Publisher'
        if len(th_cells) >= 6:
            header_row = th_cells
            break

    headers = []
    if header_row:
        for th in header_row:
            colspan = int(th.get("colspan", 1))
            label = th.get_text(" ", strip=True)
            if colspan == 2 and label.lower() == "game":
                headers.extend(["Box_Art", "Game"])
            else:
                headers.append(label)
                for extra in range(colspan - 1):
                    headers.append(f"{label}_{extra + 1}")
    return headers
```

```python
def scrape_vgchartz_page(page_number, session):
    """Scrape a single page (50 rows) from the VGChartz Game Database."""
    base_url = (
        "https://www.vgchartz.com/games/games.php"
        f"?page={page_number}&results=50&order=Sales&direction=DESC"
        "&showtotalsales=1&shownasales=1&showpalsales=1&showjapansales=1"
        "&showothersales=1&showpublisher=1&showdeveloper=1"
        "&showreleasedate=1&showcriticscore=1&showuserscore=1&showshipped=1"
    )
    headers = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/124.0.0.0 Safari/537.36"
        )
    }

    response = session.get(base_url, headers=headers, timeout=20)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")
    general_body = soup.find("div", id="generalBody")
    table = general_body.find("table") if general_body else None
    if table is None:
        raise ValueError(f"Could not locate data table on page {page_number}.")

    col_names = parse_table_headers(table)
    records = []

    for row in table.find_all("tr"):
        cells = row.find_all("td")
        # Valid game rows begin with a numeric rank in the first cell
        if len(cells) < 8 or not cells[0].get_text(strip=True).isdigit():
            continue

        row_data = {}
        for idx, cell in enumerate(cells):
            col_key = col_names[idx] if idx < len(col_names) else f"Col_{idx}"
            if col_key == "Console":
                img = cell.find("img")
                row_data[col_key] = img.get("alt", "").strip() if img else cell.get_text(strip=True)
            elif col_key == "Game":
                link = cell.find("a")
                row_data[col_key] = link.get_text(strip=True) if link else cell.get_text(strip=True)
            else:
                row_data[col_key] = cell.get_text(strip=True)

        row_data["Source_Page"] = page_number
        records.append(row_data)

    return records
```

### Execute the 5-Page Scraping Pipeline

We loop through pages 1 to 5, pausing 2 seconds between requests to respect server rate limits, and save the raw dataset to a relative path (`./data/vgchartz_raw_top250.csv`).

```python
data_dir = "./data"
os.makedirs(data_dir, exist_ok=True)
raw_csv_path = os.path.join(data_dir, "vgchartz_raw_top250.csv")

all_records = []
with requests.Session() as http_session:
    for page_num in range(1, 6):
        page_records = scrape_vgchartz_page(page_num, http_session)
        all_records.extend(page_records)
        # Polite delay between page requests to avoid HTTP 429 rate limits
        if page_num < 5:
            time.sleep(2)

raw_games_df = pd.DataFrame(all_records)
raw_games_df.to_csv(raw_csv_path, index=False)

raw_games_df.head()
```

## 3. Data Cleaning and Processing

Standardize formatted strings fomr the raw web-scraped tables:
- **Sales Columns (`Total Shipped`, `Total Sales`, `NA Sales`, `PAL Sales`, `Japan Sales`, `Other Sales`):** Values end with `"m"` (millions) or appear as `"N/A"`. Strip `"m"` and convert them to floating-point numbers.
- **Effective Global Sales:** Some newer titles are reported under `Total Shipped` when granular regional estimates are unavailable. Creating `Global_Sales_M` ensures no top-selling title is dropped.
- **Numeric Scores & Dates:** `Critic Score` and `User Score` are converted to floats, and `Release Date` is parsed into a datetime column so we can extract `Release_Year` and `Release_Decade`.

```python
def clean_sales_series(series):
    """Convert VGChartz sales strings (e.g., '30.80m', 'N/A') into numeric millions."""
    cleaned = (
        series.astype(str)
        .str.strip()
        .str.lower()
        .str.replace("m", "", regex=False)
        .replace({"n/a": np.nan, "": np.nan, "nan": np.nan})
    )
    return pd.to_numeric(cleaned, errors="coerce")
```

```python
games_df = raw_games_df.copy()

# Drop decorative box-art column if present
if "Box_Art" in games_df.columns:
    games_df = games_df.drop(columns=["Box_Art"])

# Standardize column names to snake_case-friendly labels
column_rename_map = {
    "Pos": "Rank",
    "Game": "Title",
    "Console": "Console",
    "Publisher": "Publisher",
    "Developer": "Developer",
    "Critic Score": "Critic_Score",
    "User Score": "User_Score",
    "Total Shipped": "Total_Shipped_M",
    "Total Sales": "Total_Sales_M",
    "NA Sales": "NA_Sales_M",
    "PAL Sales": "PAL_Sales_M",
    "Japan Sales": "Japan_Sales_M",
    "Other Sales": "Other_Sales_M",
    "Release Date": "Release_Date",
}
games_df = games_df.rename(columns={k: v for k, v in column_rename_map.items() if k in games_df.columns})

# Convert rank and score columns to numeric types
games_df["Rank"] = pd.to_numeric(games_df["Rank"], errors="coerce")
for score_col in ["Critic_Score", "User_Score"]:
    if score_col in games_df.columns:
        games_df[score_col] = pd.to_numeric(games_df[score_col], errors="coerce")

# Clean all sales columns into floats representing millions of units
sales_columns = [
    "Total_Shipped_M",
    "Total_Sales_M",
    "NA_Sales_M",
    "PAL_Sales_M",
    "Japan_Sales_M",
    "Other_Sales_M",
]
for col in sales_columns:
    if col in games_df.columns:
        games_df[col] = clean_sales_series(games_df[col])

# Combine Total_Sales_M with Total_Shipped_M fallback for complete global sales coverage
games_df["Global_Sales_M"] = games_df["Total_Sales_M"].fillna(games_df.get("Total_Shipped_M"))

# Parse release dates (VGChartz uses formats like '19th Nov 06')
cleaned_dates = (
    games_df["Release_Date"]
    .astype(str)
    .str.replace(r"(st|nd|rd|th)", "", regex=True)
    .str.strip()
)
games_df["Release_Date_Parsed"] = pd.to_datetime(cleaned_dates, format="%d %b %y", errors="coerce")

# Fix two-digit years parsed into the future (e.g., '85' -> 2085 should be 1985)
future_mask = games_df["Release_Date_Parsed"].dt.year > 2026
games_df.loc[future_mask, "Release_Date_Parsed"] = games_df.loc[
    future_mask, "Release_Date_Parsed"
] - pd.DateOffset(years=100)

games_df["Release_Year"] = games_df["Release_Date_Parsed"].dt.year
games_df["Release_Decade"] = (games_df["Release_Year"] // 10 * 10).astype("Int64").astype(str) + "s"
games_df.loc[games_df["Release_Year"].isna(), "Release_Decade"] = np.nan

# Save cleaned dataset to a relative path
clean_csv_path = os.path.join(data_dir, "vgchartz_clean_top250.csv")
games_df.to_csv(clean_csv_path, index=False)

games_df[["Rank", "Title", "Console", "Publisher", "Global_Sales_M", "Critic_Score", "Release_Year"]].head()
```

## 4. Exploratory Data Analysis

### Question 1: Which top 10 publishers dominate combined global sales in the top 250?

We aggregate `Global_Sales_M` by `Publisher`, rank the top 10 publishers, and calculate each publisher's percentage share of all top-250 global sales.

```python
total_top250_sales = games_df["Global_Sales_M"].sum()

publisher_summary = (
    games_df.groupby("Publisher", as_index=False)
    .agg(
        Total_Sales_M=("Global_Sales_M", "sum"),
        Game_Count=("Title", "count"),
        Avg_Sales_M=("Global_Sales_M", "mean"),
    )
    .sort_values("Total_Sales_M", ascending=False)
    .head(10)
    .reset_index(drop=True)
)
publisher_summary["Market_Share_Pct"] = (publisher_summary["Total_Sales_M"] / total_top250_sales) * 100

publisher_summary.round(2)
```

```python
# Extract key metrics dynamically for a conclusion-driven title
top_publisher = publisher_summary.loc[0, "Publisher"]
top_pub_sales = publisher_summary.loc[0, "Total_Sales_M"]
top_pub_share = publisher_summary.loc[0, "Market_Share_Pct"]

fig, ax = plt.subplots(figsize=(11, 6))
sns.barplot(
    data=publisher_summary,
    x="Total_Sales_M",
    y="Publisher",
    hue="Publisher",
    palette="viridis",
    legend=False,
    ax=ax,
)

# Add data labels showing both total millions sold and market share percentage
for bar, (_, row) in zip(ax.patches, publisher_summary.iterrows()):
    width = bar.get_width()
    ax.text(
        width + (top_pub_sales * 0.015),
        bar.get_y() + bar.get_height() / 2,
        f"{width:.1f}M ({row['Market_Share_Pct']:.1f}%)",
        va="center",
        ha="left",
        fontsize=9.5,
        fontweight="bold",
        color="#222222",
    )

ax.set_title(
    f"Publisher Sales in the Top 250 Video Games",
    fontsize=13,
    fontweight="bold",
    pad=14,
)
ax.set_xlabel("Combined Global Sales (Millions of Units)", fontsize=11)
ax.set_ylabel("Publisher", fontsize=11)
ax.set_xlim(0, top_pub_sales * 1.24)
sns.despine(left=True, bottom=True)
plt.tight_layout()
plt.show()
```

**Plot Interpretation (Question 1):**  
The horizontal bar chart displays the 10 highest-grossing publishers in the VGChartz Top 250. The leading publisher outpaces every competitor by a wide margin in both total unit sales and overall market share, highlighting how heavily blockbuster franchises concentrate revenue among a small group of major publishers.

### Question 2: How Do Regional Sales Proportions Differ Across the Top 5 Consoles?

To compare regional demand across major consoles, we filter for games with complete regional data (`NA_Sales_M`, `PAL_Sales_M`, `Japan_Sales_M`), identify the top 5 consoles by combined regional sales, and compute each region's percentage share of that console's sales.

```python
regional_df = games_df.dropna(subset=["NA_Sales_M", "PAL_Sales_M", "Japan_Sales_M"]).copy()

console_regional = (
    regional_df.groupby("Console", as_index=False)[["NA_Sales_M", "PAL_Sales_M", "Japan_Sales_M"]]
    .sum()
)
console_regional["Regional_Total_M"] = (
    console_regional["NA_Sales_M"] + console_regional["PAL_Sales_M"] + console_regional["Japan_Sales_M"]
)

top5_consoles = (
    console_regional.sort_values("Regional_Total_M", ascending=False)
    .head(5)
    .reset_index(drop=True)
)

# Calculate percentage share per region for each console
top5_consoles["North America (%)"] = (top5_consoles["NA_Sales_M"] / top5_consoles["Regional_Total_M"]) * 100
top5_consoles["PAL / Europe (%)"] = (top5_consoles["PAL_Sales_M"] / top5_consoles["Regional_Total_M"]) * 100
top5_consoles["Japan (%)"] = (top5_consoles["Japan_Sales_M"] / top5_consoles["Regional_Total_M"]) * 100

top5_consoles[["Console", "Regional_Total_M", "North America (%)", "PAL / Europe (%)", "Japan (%)"]].round(1)
```

```python
# Reshape data to long format for grouped Seaborn bar plotting
regional_melted = top5_consoles.melt(
    id_vars="Console",
    value_vars=["North America (%)", "PAL / Europe (%)", "Japan (%)"],
    var_name="Region",
    value_name="Sales_Share_Pct",
)

# Determine the console with the highest Japan share for a data-backed title
highest_jp_row = top5_consoles.loc[top5_consoles["Japan (%)"].idxmax()]
highest_na_row = top5_consoles.loc[top5_consoles["North America (%)"].idxmax()]

fig, ax = plt.subplots(figsize=(11, 6))
sns.barplot(
    data=regional_melted,
    x="Console",
    y="Sales_Share_Pct",
    hue="Region",
    palette="Set2",
    ax=ax,
)

# Add percentage data labels above each bar
for container in ax.containers:
    ax.bar_label(container, fmt="%.1f%%", padding=3, fontsize=8.5, fontweight="bold")

ax.set_title(
    f"Regional Shares of Console Sales",
    fontsize=12,
    fontweight="bold",
    pad=14,
)
ax.set_xlabel("Console Platform", fontsize=11)
ax.set_ylabel("Share of Combined Regional Sales (%)", fontsize=11)
ax.set_ylim(0, regional_melted["Sales_Share_Pct"].max() * 1.18)
ax.legend(title="Region", frameon=True)
sns.despine()
plt.tight_layout()
plt.show()
```

**Plot Interpretation (Question 2):**  
Comparing regional sales shares reveals distinct geographic strengths by platform. Western markets (North America and PAL/Europe) generate the majority of top-250 software sales across consoles, while handheld and Nintendo platforms capture a noticeably larger share of sales in Japan than competing systems.

### Question 3: Is there a positive correlation between critic score and global sales?

We filter to top-250 games with valid `Critic_Score` and `Global_Sales_M` values, compute the Pearson correlation coefficient, and use a Seaborn regression scatter plot to visualize the relationship, with data labels on the five highest-selling scored titles.

```python
scored_games = games_df.dropna(subset=["Critic_Score", "Global_Sales_M"]).copy()
pearson_r = scored_games["Critic_Score"].corr(scored_games["Global_Sales_M"])

correlation_stats = pd.Series(
    {
        "Scored_Games_Count": len(scored_games),
        "Mean_Critic_Score": round(scored_games["Critic_Score"].mean(), 2),
        "Mean_Global_Sales_M": round(scored_games["Global_Sales_M"].mean(), 2),
        "Pearson_r": round(pearson_r, 3),
    },
    name="Metric",
)
correlation_stats
```

```python
# Classify correlation strength programmatically for the plot title
abs_r = abs(pearson_r)
if abs_r >= 0.5:
    strength_label = "Strong"
elif abs_r >= 0.3:
    strength_label = "Moderate"
else:
    strength_label = "Weak"
direction_label = "Positive" if pearson_r >= 0 else "Negative"

fig, ax = plt.subplots(figsize=(11, 6))
sns.regplot(
    data=scored_games,
    x="Critic_Score",
    y="Global_Sales_M",
    scatter_kws={"alpha": 0.65, "s": 60, "color": "#3b75af"},
    line_kws={"color": "#d95f02", "linewidth": 2.2},
    ax=ax,
)

# Annotate the top 5 highest-selling games with data labels
top5_scored = scored_games.nlargest(5, "Global_Sales_M")
for _, row in top5_scored.iterrows():
    short_title = row["Title"] if len(row["Title"]) <= 24 else row["Title"][:21] + "..."
    ax.annotate(
        f"{short_title} ({row['Global_Sales_M']:.1f}M)",
        xy=(row["Critic_Score"], row["Global_Sales_M"]),
        xytext=(6, 6),
        textcoords="offset points",
        fontsize=8.5,
        fontweight="bold",
        color="#222222",
    )

ax.set_title(
    f"Positive Correlation Between Critic Score and Top-250 Sales",
    fontsize=13,
    fontweight="bold",
    pad=14,
)
ax.set_xlabel("VGChartz Critic Score (Out of 10)", fontsize=11)
ax.set_ylabel("Global Sales (Millions of Units)", fontsize=11)
sns.despine()
plt.tight_layout()
plt.show()
```

**Plot Interpretation (Question 3):**  
Because every title in this dataset is already a commercial blockbuster (top 250 all-time), critic scores cluster heavily toward the high end of the scale, and massive pack-in or casual hits can outsell critically higher-rated games. The trendline and Pearson $r$ value quantify how weakly or strongly review scores alone predict sales volume among elite sellers.

### Question 4: Which consoles achieve the highest average sales per title, and which decade produced the most hits?

1. **Console Performance:** Mean, median, total global sales, and entry count for consoles with at least 5 titles in the top 250.
2. **Release Decade Trends:** Number of top-250 games and average global sales grouped by release decade.

```python
console_performance = (
    games_df.groupby("Console", as_index=False)
    .agg(
        Title_Count=("Title", "count"),
        Mean_Sales_M=("Global_Sales_M", "mean"),
        Median_Sales_M=("Global_Sales_M", "median"),
        Total_Sales_M=("Global_Sales_M", "sum"),
    )
    .query("Title_Count >= 5")
    .sort_values("Mean_Sales_M", ascending=False)
    .reset_index(drop=True)
)

console_performance.round(2)
```

```python
decade_performance = (
    games_df.dropna(subset=["Release_Decade"])
    .groupby("Release_Decade", as_index=False)
    .agg(
        Title_Count=("Title", "count"),
        Mean_Sales_M=("Global_Sales_M", "mean"),
        Total_Sales_M=("Global_Sales_M", "sum"),
    )
    .sort_values("Release_Decade")
    .reset_index(drop=True)
)

decade_performance.round(2)
```

```python
top_avg_console = console_performance.loc[0, "Console"]
top_avg_console_sales = console_performance.loc[0, "Mean_Sales_M"]
most_prolific_decade = decade_performance.loc[decade_performance["Title_Count"].idxmax()]

conclusions_df = pd.DataFrame(
    [
        {
            "Question": "1. Top Publisher",
            "Programmatic_Finding": (
                f"{top_publisher} leads with {top_pub_sales:.2f}M units "
                f"({top_pub_share:.1f}% of top-250 sales)."
            ),
        },
        {
            "Question": "2. Regional Demand",
            "Programmatic_Finding": (
                f"{highest_na_row['Console']} has the highest NA share ({highest_na_row['North America (%)']:.1f}%), "
                f"while {highest_jp_row['Console']} leads in Japan share ({highest_jp_row['Japan (%)']:.1f}%)."
            ),
        },
        {
            "Question": "3. Critic Score Correlation",
            "Programmatic_Finding": (
                f"Pearson r = {pearson_r:.3f} across {len(scored_games)} scored titles "
                f"({strength_label.lower()} {direction_label.lower()} correlation)."
            ),
        },
        {
            "Question": "4. Console & Decade Leaders",
            "Programmatic_Finding": (
                f"{top_avg_console} leads average sales per hit ({top_avg_console_sales:.2f}M/title); "
                f"the {most_prolific_decade['Release_Decade']} produced the most top-250 games "
                f"({int(most_prolific_decade['Title_Count'])} titles)."
            ),
        },
    ]
)

conclusions_df
```
